# 0. Writing Useful Python Comments

Python uses `#` for comments. Comments are ignored when the code runs, but they
help a reader understand the analysis.

- A **full-line comment** appears above the relevant code. Use it to explain
  purpose, reasoning, an assumption, or an engineering unit.
- A short **inline comment** follows a statement. Use it sparingly for a brief
  clarification.
- A longer note uses several consecutive `#` lines. Python does not have a
  dedicated block-comment syntax.
- A **docstring** is a triple-quoted string inside a function. It documents the
  function's purpose and is not an ordinary comment.
- A `# TODO:` comment marks code that still needs to be written.

Prefer comments that explain **why**. Avoid comments that only repeat an
obvious statement such as `count = 0  # Set count to zero`.

```python
# Full-line comment: convert rotation speed to cycles per second.
example_speed_rpm = 1_800  # Inline comment: hypothetical operating point.
example_speed_hz = example_speed_rpm / 60
print("Rotation rate:", example_speed_hz, "Hz")


def convert_mm_min_to_mm_s(feed_rate_mm_min):
    '''Return feed rate converted from mm/min to mm/s.'''
    # Divide by 60 because one minute contains 60 seconds.
    return feed_rate_mm_min / 60


print("Feed rate:", convert_mm_min_to_mm_s(300.0), "mm/s")
```

# 8. Loops

A `for` loop repeats a block once for each item. `range()` generates integer
indices, `enumerate()` provides an index and an item, and `zip()` pairs items
from equally sized sequences.

```python
# Repeat an action for each sensor.
sensors = ["Force", "Vibration", "Temperature"]

for sensor in sensors:
    print("Checking:", sensor)

print("Check complete")
```

```python
# Use range and enumerate for numbered output.
for cut_number in range(1, 4):
    print("Processing cut", cut_number)

for index, sensor in enumerate(sensors, start=1):
    print(index, sensor)
```

```python
# Pair related lists with zip.
time_s = [0, 1, 2, 3]
pressure_kPa = [101, 118, 133, 141]

for time_value, pressure_value in zip(time_s, pressure_kPa):
    print(f"t = {time_value} s, pressure = {pressure_value} kPa")
```

# 9. Functions

A function gives a reusable name to a calculation. Parameters receive input
values, and `return` sends a result back to the calling code. Keep a function
focused on one clear task.

```python
# Define and call a unit-conversion function.
def rpm_to_hz(speed_rpm):
    '''Return rotation rate in cycles per second.'''
    return speed_rpm / 60


rotation_rate_hz = rpm_to_hz(3_600)
print("Rotation rate:", rotation_rate_hz, "Hz")
```

```python
# A function can use more than one parameter.
def calculate_mean(values):
    '''Return the arithmetic mean of a nonempty list.'''
    return sum(values) / len(values)


current_measurements_A = [5.2, 5.5, 5.4, 5.7]
mean_current_A = calculate_mean(current_measurements_A)
print("Mean current:", round(mean_current_A, 2), "A")
```

```python
# TRY IT: Modify the function to use a different threshold.
def classify_temperature(temperature_C, warning_limit_C=80.0):
    '''Classify one temperature using a course-example threshold.'''
    if temperature_C >= warning_limit_C:
        return "warning"
    return "normal"


print(classify_temperature(76.0))
print(classify_temperature(84.0))
print(classify_temperature(76.0, warning_limit_C=75.0))
```

**Example answer:** The three calls return normal, warning, and warning. The last call supplies 75.0 °C instead of the default 80.0 °C; therefore 76.0 °C changes category.

# 11. Integrated Mini-Practice

The following example combines a list-based table, a loop, a condition, a
dictionary, a function, and formatted output. Notice how each Python tool has a
specific role in the workflow.

```python
# Analyze a short set of pump measurements.
pump = {
    "name": "Pump P-04",
    "pressure_limit_kPa": 300,
}

# Each row contains [time_s, pressure_kPa].
pump_measurements = [
    [0, 220],
    [1, 248],
    [2, 276],
    [3, 305],
    [4, 318],
]

def is_pressure_warning(pressure_kPa, limit_kPa):
    '''Return True when pressure is above the course-example limit.'''
    return pressure_kPa > limit_kPa


warning_times_s = []

for row in pump_measurements:
    time_value_s = row[0]
    pressure_value_kPa = row[1]
    if is_pressure_warning(pressure_value_kPa, pump["pressure_limit_kPa"]):
        warning_times_s.append(time_value_s)

print(f"{pump['name']} pressure limit: {pump['pressure_limit_kPa']} kPa")
print("Warning times (s):", warning_times_s)
print("Number of warnings:", len(warning_times_s))
```

### Mini-practice interpretation

At the original 300 kPa limit, pressures at 3 and 4 s exceed the limit.
Lowering it to 275 kPa also flags 2 s: the warning count increases from 2 to 3.
The measurements remain unchanged; only the classification rule changes.
The function uses `>`, so a pressure exactly equal to the limit is not a warning.
These flags do not establish a physical fault.
