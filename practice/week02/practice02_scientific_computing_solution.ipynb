`axis=0` reduces rows, leaving one value per column. `axis=1` reduces columns,

A row average is only a derived two-sensor arithmetic average, not necessarily
the physical temperature of an entire part.

**Worked example:** extract the left channel without changing the table.

````python
# Rows are samples; columns are left and right temperatures in degrees Celsius.
B_temp_C = np.array([[20.0, 24.0], [22.0, 24.0],
                     [24.0, 26.0], [26.0, 30.0]])
print("Shape (samples, channels):", B_temp_C.shape)
print("Left channel (°C):", B_temp_C[:, 0])
````

# Module D — Sample indices and physical time

The explicitly assumed sampling rate here is 4 Hz (4 samples per second),
starting at time zero. Samples have indices `0, ..., N-1`; `np.arange(N)`
creates these indices. Divide by `fs_hz` for seconds.

For uniform samples, sample spacing is `1/fs_hz`, the nominal N-sample record
duration is `N/fs_hz`, and the last timestamp is `(N-1)/fs_hz`. The first-to-last
timestamp span has N-1 intervals. These are different definitions, not conflicting
measurements. No sampling rate is inferred from the signal values.

**Worked example:** construct timestamps for a separate three-sample example.

````python
# Synthetic pressure readings with an explicitly assigned 4 Hz sampling rate.
D_pressure_kPa = np.array([100.0, 102.0, 101.0, 103.0, 102.0, 100.0])
D_fs_hz = 4
print("Three-sample example times (s):", np.arange(3) / D_fs_hz)
````

````python
# Use indices starting at zero and the stated sampling rate.
D1_N = len(D_pressure_kPa)
D1_time_s = np.arange(D1_N) / D_fs_hz
print("Times (s):", D1_time_s)
print("N:", D1_N, "Spacing (s):", 1 / D_fs_hz)
print("Nominal duration (s):", D1_N / D_fs_hz)
print("Last timestamp (s):", D1_time_s[-1])
````

**One possible approach / explanation:**

The times are [0, 0.25, 0.5, 0.75, 1, 1.25] s. N=6, spacing=0.25 s, nominal duration=1.5 s, and last timestamp=1.25 s. Six samples occupy six nominal sampling intervals, but only five intervals separate first and last timestamps.

# Module E — Mean, sample SD, offsets, and units

Mean describes average level. Sample standard deviation (SD) describes
spread about that mean: square deviations, sum them, divide by n-1, then take
the square root. Mean and SD have the original measurement unit. Use
`np.std(x, ddof=1)` for sample SD; NumPy's default `ddof=0` divides by n.
At least two available values are needed for sample SD. The formula alone
does not establish independent sampling or justify inference about a machine.

**Worked example:** compute both summaries for three synthetic current readings.

````python
# Synthetic current sequences; they are not fault-labeled measurements.
E_current_A = np.array([2.0, 4.0, 6.0])
E_flat_A = np.array([4.0, 4.0, 4.0])
print("Mean current (A):", np.mean(E_current_A))
print("Sample SD (A):", np.std(E_current_A, ddof=1))
````

## E1 — Compare: same mean, different variability

Predict whether `E_flat_A` and `E_current_A` have equal means and equal
sample SDs. Calculate both and explain what the mean alone hides.

````python
# Compare average level and spread separately.
print("Changing sequence: mean / sample SD (A):",
      np.mean(E_current_A), np.std(E_current_A, ddof=1))
print("Flat sequence: mean / sample SD (A):",
      np.mean(E_flat_A), np.std(E_flat_A, ddof=1))
````

**One possible approach / explanation:**

Both means are 4 A, but sample SDs are 0 A for the flat sequence and 2 A for [2, 4, 6]. Equal mean does not imply equal variation; neither sequence is a validated healthy/faulty example.

## E2 — Change: constant offset

Add 10 A to every value of `E_current_A` in a new array. Predict and
calculate the new mean and sample SD. Explain using deviations from the mean.
Do not interpret the synthetic offset as a known real sensor bias.

````python
# A constant shift preserves deviations from the shifted mean.
E2_shifted_A = E_current_A + 10
print("Shifted values (A):", E2_shifted_A)
print("Mean / sample SD (A):", np.mean(E2_shifted_A), np.std(E2_shifted_A, ddof=1))
print("Deviations (A):", E2_shifted_A - np.mean(E2_shifted_A))
````

**One possible approach / explanation:**

The mean increases from 4 to 14 A, but sample SD stays 2 A. Deviations remain [-2, 0, 2] A because the mean shifts by the same offset. This arithmetic property does not identify the cause of an offset in real data.

## E3 — Change: A to mA

Convert `E_current_A` to mA using 1000 mA = 1 A. Compute mean and sample

````python
# Apply the same unit conversion to every measurement.
E3_current_mA = E_current_A * 1000
print("Current (mA):", E3_current_mA)
print("Mean / sample SD (mA):",
      np.mean(E3_current_mA), np.std(E3_current_mA, ddof=1))
````

## E4 — Debug: the SD convention

An analyst reports `np.std(E_current_A)` as sample SD. Calculate that
value and the result with `ddof=1`. Check the latter by calculating squared
deviations, dividing their sum by n-1, and using `np.sqrt`. Explain the difference.

````python
# Check the denominator rather than assuming library defaults agree.
E4_deviation_A = E_current_A - np.mean(E_current_A)
E4_manual_A = np.sqrt(np.sum(E4_deviation_A ** 2) / (len(E_current_A) - 1))
print("Default SD (A):", np.std(E_current_A))
print("Sample SD (A):", np.std(E_current_A, ddof=1))
print("Manual sample SD (A):", E4_manual_A)
````

**One possible approach / explanation:**

The default result is about 1.633 A (divisor 3); sample SD is 2 A (divisor 2). Squared deviations sum to 8 A², so sqrt(8/2)=2 A. Different divisors answer different statistical conventions; specify ddof=1 for this practice.

## F2 — Compare: shared panel ranges

Make two panels with shared x and y axes, one record per panel, using
0–6 s and 18–28 °C in both. Compare their numerical temperature ranges.
Why might separately autoscaled panels obscure this comparison?

````python
# Match display ranges before comparing panel heights.
fig, axes = plt.subplots(1, 2, figsize=(9, 3), sharex=True, sharey=True)
axes[0].plot(F_time_A_s, F_temp_A_C, marker="o", label="Record A")
axes[1].plot(F_time_B_s, F_temp_B_C, marker="s", label="Record B")
axes[0].set_title("Synthetic record A")
axes[1].set_title("Synthetic record B")
for ax in axes:
    ax.set_xlim(0, 6)
    ax.set_ylim(18, 28)
    ax.set_xlabel("Time from record start (s)")
    ax.set_ylabel("Temperature (°C)")
    ax.legend()
    ax.grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
print("Ranges A / B (°C):", np.max(F_temp_A_C) - np.min(F_temp_A_C),
      np.max(F_temp_B_C) - np.min(F_temp_B_C))
````

**One possible approach / explanation:**

A spans 20–21 °C (range 1 °C); B spans 24–26 °C (range 2 °C). Separate autoscaling can make both traces fill similar panel heights. Shared ranges reveal the difference in levels and spans without altering the samples.

## F5 — Optional challenge: equal mean and SD, different order
