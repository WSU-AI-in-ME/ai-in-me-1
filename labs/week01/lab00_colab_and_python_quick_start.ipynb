## Python comments

Python ignores text after `#`. Use comments to explain the purpose of a step,
an assumption, or a unit that may not be obvious from the code. A full-line
comment appears on its own line. A short inline comment follows code on the
same line. Do not comment every obvious statement.

```python
# A full-line comment explains the purpose of the next statement.
machine_mode = "Training"  # An inline comment adds a short clarification.
print("Machine mode:", machine_mode)
```

## A variable and arithmetic

A variable gives a value a name. The next cell stores a spindle speed and
calculates twice that value.

```python
# [GUIDED] Edit the value, then run the cell again.
spindle_speed_rpm = 8_000
double_speed_rpm = spindle_speed_rpm * 2
print("Spindle speed:", spindle_speed_rpm, "rpm")
print("Twice the spindle speed:", double_speed_rpm, "rpm")
```

## A short list and indexing

Python uses index `0` for the first item. Index `-1` selects the last item.

```python
# [GUIDED] Select items from a short list.
channels = ["Force X", "Vibration X", "AE-RMS"]
print("First channel:", channels[0])
print("Last channel:", channels[-1])
```

## Runtime state

The Colab runtime remembers variables from previously executed cells. Run the
next cell several times and watch the number increase. The cell can also use
`machine_status`, which was created earlier in the notebook.

```python
# [GUIDED] Run this cell more than once.
run_count = globals().get("run_count", 0) + 1
print("Number of executions:", run_count)
print("Machine status:", machine_status)
```
