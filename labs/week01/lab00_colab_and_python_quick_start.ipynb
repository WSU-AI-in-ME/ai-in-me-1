Python ignores text after `#`. Use comments to explain the purpose of a step,
an assumption, or a unit that may not be obvious from the code. A full-line
comment appears on its own line. A short inline comment follows code on the
same line. Do not comment every obvious statement.

```python
# A full-line comment explains the purpose of the next statement.
machine_mode = "Training"  # An inline comment adds a short clarification.
print("Machine mode:", machine_mode)
```
