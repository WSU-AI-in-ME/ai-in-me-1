## [GUIDED] Python comments and readable code

Python uses `#` for comments.

- A **full-line comment** appears above the code and explains purpose,
  reasoning, an assumption, or an engineering unit.
- A short **inline comment** follows a statement. Use inline comments
  sparingly so code remains readable.
- For a longer explanation, use several consecutive `#` lines. Python has no
  separate block-comment syntax.
- A **docstring** is a triple-quoted string placed inside a function to
  document what the function does. It is documentation, not an ordinary
  comment.
- `# TODO:` is a full-line comment that marks work you must complete.

Comments should explain **why** or clarify engineering meaning. Avoid comments
that merely restate obvious Python syntax.

Markdown cells are for instructions, discussion, and written responses. Python
comments belong inside code cells.

# Part 4 — Basic Visualization with Matplotlib

## [GUIDED] Tool-wear trend from Python lists

The x-axis contains the independent variable, cut number. The y-axis contains
the measured quantity and its physical unit. A descriptive title, marker,
line, grid, legend, `tight_layout()`, and `plt.show()` make the figure readable.

```python
# Import Matplotlib using its conventional pyplot alias.
import matplotlib.pyplot as plt

# These two lists contain paired x- and y-values.
cut_number = [1, 2, 3, 4, 5]
tool_wear_um = [18, 25, 33, 46, 61]

# Build one labeled engineering figure from the paired lists.
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(cut_number, tool_wear_um, marker="o", linewidth=2, label="Tool wear")
ax.set_xlabel("Cut number")
ax.set_ylabel("Tool wear (µm)")
ax.set_title("Tool Wear Progression")
ax.grid(True, alpha=0.3)
ax.legend()
fig.tight_layout()
plt.show()
```
