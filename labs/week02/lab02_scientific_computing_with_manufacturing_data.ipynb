# Part 3 — First and Last Machining Records

## [GUIDED] Inspect two complete records

Use cutter `c1` cuts 1 and 315. The seven columns are Force X/Y/Z in N,
Vibration X/Y/Z in g, and AE-RMS in V. Here **g is an acceleration unit based on
standard gravity, not gram: 1 g = 9.80665 m/s²**. Each row holds simultaneous channel samples.
The [official challenge](https://phmsociety.org/phm_competition/2010-phm-society-conference-data-challenge/)
documents **50,000 Hz/channel**. The CSV files do not store a sampling rate or timestamps.

**Interpretation limits, before comparing:** Cut 1 is a first sequence position,
not an unworn or established stable-cutting baseline. These are complete records,
with different lengths and no alignment to the same cutting phase. Engagement
annotations are unavailable. Included operating states and their proportions can
affect whole-record statistics, which are not cutting-only statistics.
Differences cannot establish wear causation, chatter, or failure.

```python
# Load both existing records; keep every recorded sample.
first_record = pd.read_csv(phm_url + "c1_cut001.csv")
last_record = pd.read_csv(phm_url + "c1_cut315.csv")
print("Cut 1 shape:", first_record.shape)
print("Cut 315 shape:", last_record.shape)
print("Cut 1 columns:", first_record.columns.tolist())
print("Cut 315 columns:", last_record.columns.tolist())
display(pd.DataFrame({"Cut 1 missing": first_record.isna().sum(),
                      "Cut 315 missing": last_record.isna().sum()}))
```

## [GUIDED] From samples to seconds and statistics

For `N` samples at `fs_hz`, `np.arange(N)` returns indices `0` through `N-1`.
Dividing by samples/second gives time in seconds. The nominal duration is `N/fs_hz`;
the last sample timestamp is `(N-1)/fs_hz`. They differ by one sampling interval.
Use **each record's own N**. `.to_numpy()` converts a selected pandas column to an array.

The **mean** summarizes the signal's average signed level. The **standard deviation**
summarizes spread around that mean and has the same unit as the signal. Positive
and negative values can cancel in the mean while variation remains substantial.
In the [official `np.std` documentation](https://numpy.org/doc/stable/reference/generated/numpy.std.html),
find `ddof`: its default is `0`; this Lab uses **`ddof=1`**, so the squared-deviation
sum is divided by `N-1` before taking the square root. This is sample SD; it does
not make time-correlated samples independent experiments.

The provided `pd.DataFrame({...})` below collects equal-length lists into a compact
table; dictionary keys become column names. `.round()` returns a new object
containing rounded values. Here `display(summary.round(...))` displays that new
object without changing the original summary or the signal arrays.

```python
# Convert the documented rate and each sample count into physical time.
fs_hz = 50_000
n_first = first_record.shape[0]
n_last = last_record.shape[0]
time_first_s = np.arange(n_first) / fs_hz
time_last_s = np.arange(n_last) / fs_hz
duration_first_s = n_first / fs_hz
duration_last_s = n_last / fs_hz
last_time_first_s = (n_first - 1) / fs_hz
last_time_last_s = (n_last - 1) / fs_hz

# Convert Force X columns to arrays and summarize each complete record.
force_first = first_record["force_x_N"].to_numpy()
force_last = last_record["force_x_N"].to_numpy()
force_summary = pd.DataFrame({
    "Cut": [1, 315], "Samples": [n_first, n_last],
    "Duration (s)": [duration_first_s, duration_last_s],
    "Last time (s)": [last_time_first_s, last_time_last_s],
    "Mean (N)": [np.mean(force_first), np.mean(force_last)],
    "Sample SD (N)": [np.std(force_first, ddof=1), np.std(force_last, ddof=1)],
})
display(force_summary.round(6))
```

## [GUIDED] Axis scaling A — independent y-axis limits

`plt.subplots(1, 2, ...)` returns one figure and two axes: `axes[0]` and `axes[1]`.
First, use automatic, independent y-axis limits. Keep a common time range so
the next comparison changes **only the y-axis scaling**, not the data or time range.
Independent y limits can reveal small waveform shapes, but similar visual heights
on different scales do not imply similar amplitudes.

```python
# Same data and time range as the next figure; each y-axis autoscales independently.
fig_auto, axes_auto = plt.subplots(1, 2, figsize=(10, 3.2), sharex=True, sharey=False)
axes_auto[0].plot(time_first_s, force_first, linewidth=0.5)
axes_auto[1].plot(time_last_s, force_last, linewidth=0.5, color="tab:orange")
axes_auto[0].set_title("Cut 1 — Force X")
axes_auto[1].set_title("Cut 315 — Force X")
axes_auto[0].set_xlabel("Time (s)")
axes_auto[1].set_xlabel("Time (s)")
axes_auto[0].set_ylabel("Force X (N)")
axes_auto[0].set_xlim(0, max(duration_first_s, duration_last_s))
axes_auto[0].grid(True, alpha=0.3)
axes_auto[1].grid(True, alpha=0.3)
fig_auto.tight_layout()
plt.show()
```

## [GUIDED] Axis scaling B — common y-axis limits

`sharex=True, sharey=True` links both scales. The provided min/max code finds a
range covering **both records** and adds a small margin. Set limits on one shared
axis to apply them to both. The shorter recording ends naturally; blank space
after its last sample is not missing data to fill with zeros.

```python
# Use common limits so the two complete records can be compared fairly.
force_low = min(np.min(force_first), np.min(force_last))
force_high = max(np.max(force_first), np.max(force_last))
force_pad = 0.05 * (force_high - force_low)
fig, axes = plt.subplots(1, 2, figsize=(10, 3.2), sharex=True, sharey=True)
axes[0].plot(time_first_s, force_first, linewidth=0.5)
axes[1].plot(time_last_s, force_last, linewidth=0.5, color="tab:orange")
axes[0].set_title("Cut 1 — Force X")
axes[1].set_title("Cut 315 — Force X")
axes[0].set_xlabel("Time (s)")
axes[1].set_xlabel("Time (s)")
axes[0].set_ylabel("Force X (N)")
axes[0].set_xlim(0, max(duration_first_s, duration_last_s))
axes[0].set_ylim(force_low - force_pad, force_high + force_pad)
axes[0].grid(True, alpha=0.3)
axes[1].grid(True, alpha=0.3)
fig.tight_layout()
plt.show()
```

## [GUIDED] Full signal and one 20 ms zoom

Choose approximately 20 ms around the **time midpoint of Cut 315**, away from the
record boundaries by a reproducible rule. The midpoint is not guaranteed to be
stable cutting. This is one slicing example, not repeated segmentation,
cutting-state detection or chatter diagnosis.

Multiplying seconds by `fs_hz` converts time to sample index. `round()` chooses
the nearest sample boundary; `int()` gives an integer for slicing. Stop indices
are exclusive, as in Part 1. Slice the time and force arrays with the **same indices**;
keep timestamps relative to the original record start. A simple `assert` stops
the cell if its condition is false. Check index bounds **before slicing**, then
check both array lengths against the expected **1,000 samples** before displaying
the last included timestamp.

```python
# Select approximately 20 ms around the last record's time midpoint.
zoom_width_s = 0.020
zoom_center_s = duration_last_s / 2
zoom_start_s = zoom_center_s - zoom_width_s / 2
zoom_stop_s = zoom_center_s + zoom_width_s / 2
zoom_start_index = int(round(zoom_start_s * fs_hz))
zoom_stop_index = int(round(zoom_stop_s * fs_hz))
# Check bounds before selecting samples.
zoom_bounds_ok = 0 <= zoom_start_index < zoom_stop_index <= n_last
assert zoom_bounds_ok, "Choose start/stop indices within the record."
zoom_time_s = time_last_s[zoom_start_index:zoom_stop_index]
zoom_force_N = force_last[zoom_start_index:zoom_stop_index]
# Check the 20 ms selection before accessing its final timestamp.
zoom_expected_samples = int(round(zoom_width_s * fs_hz))
zoom_lengths_ok = len(zoom_time_s) == len(zoom_force_N) == zoom_expected_samples == 1000
assert zoom_lengths_ok, "Time and force must both contain the expected 1,000 samples."
print("Valid index bounds:", zoom_bounds_ok, "Expected sample count:", zoom_lengths_ok)
print("Indices [start, stop):", zoom_start_index, zoom_stop_index)
print("Selected interval [start, stop) (s):", zoom_start_index / fs_hz, zoom_stop_index / fs_hz)
print("Selected samples:", len(zoom_force_N), "Last included time (s):", zoom_time_s[-1])
```

```python
# Show the selection on the full record and retain original timestamps in the zoom.
fig_zoom, axes_zoom = plt.subplots(1, 2, figsize=(10, 3.2), sharey=True)
axes_zoom[0].plot(time_last_s, force_last, linewidth=0.5)
axes_zoom[0].axvspan(zoom_start_index / fs_hz, zoom_stop_index / fs_hz,
                   color="tab:red", alpha=0.45, label="Selected 20 ms")
axes_zoom[0].set_xlim(0, duration_last_s)
axes_zoom[0].set_title("Cut 315 — full Force X")
axes_zoom[0].set_ylabel("Force X (N)")
axes_zoom[0].legend()
axes_zoom[1].plot(zoom_time_s, zoom_force_N, linewidth=0.8)
axes_zoom[1].set_xlim(zoom_start_index / fs_hz, zoom_stop_index / fs_hz)
axes_zoom[1].set_title("Cut 315 — midpoint zoom")
axes_zoom[0].set_xlabel("Time from record start (s)")
axes_zoom[1].set_xlabel("Time from record start (s)")
axes_zoom[0].grid(True, alpha=0.3)
axes_zoom[1].grid(True, alpha=0.3)
fig_zoom.tight_layout()
plt.show()
```

## Vibration histogram

The records have different sample counts, so raw bin counts reflect recording
length as well as distribution shape. The supplied example uses **the same bin
edges** and `density=True` for both records. `np.linspace(low, high, 41)` makes
41 equally spaced edges for 40 bins covering both arrays.

Density is **probability density**, not the probability of each bar. A bar's
height times its bin width gives the fraction of samples in that bin; the total
area is one. This normalization of histogram area does not normalize the signal
amplitudes. See [Matplotlib `hist`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.hist.html).

**Optional question:** What information about timing is lost in a histogram?
It omits sample order and when events occurred; different time sequences can
have the same histogram. This is not a test for Gaussianity or stationarity.

```python
# Optional only: use identical edges and density for different record lengths.
RUN_OPTIONAL_VIBRATION_HISTOGRAM = False
if RUN_OPTIONAL_VIBRATION_HISTOGRAM:
    hist_low_g = min(np.min(vibration_first), np.min(vibration_last))
    hist_high_g = max(np.max(vibration_first), np.max(vibration_last))
    hist_edges_g = np.linspace(hist_low_g, hist_high_g, 41)
    fig_hist, ax_hist = plt.subplots(figsize=(9, 3.2))
    density_first, edges_first, bars_first = ax_hist.hist(
        vibration_first, bins=hist_edges_g, density=True, alpha=0.5, label="Cut 1")
    density_last, edges_last, bars_last = ax_hist.hist(
        vibration_last, bins=hist_edges_g, density=True, alpha=0.5, label="Cut 315")
    ax_hist.set_xlabel("Vibration X (g)")
    ax_hist.set_ylabel("Probability density (1/g)")
    ax_hist.set_title("Optional: whole-record Vibration X distributions")
    ax_hist.legend()
    ax_hist.grid(True, alpha=0.3)
    fig_hist.tight_layout()
    plt.show()
```
