### ME 5995 — Fundamentals of Manufacturing Data Science
Wayne State University

[Course Repository](https://github.com/WSU-AI-in-ME/ai-in-me-1)

Homework is independent, graded work. This index links to student-facing
homework starter files. Canvas and the
course syllabus are authoritative for assignment requirements, due dates,
rubrics, submission instructions, and course policies.

Reference Solution PDFs are provided through Canvas after the applicable
submission deadline, including approved extensions. Executable Homework
solutions are not published in GitHub.

Coding homework will normally use a `.ipynb` starter notebook so that code,
figures, and short engineering responses remain in one reproducible file.
Supplemental `.md` files may explain setup, data provenance, or repository
structure, but they do not replace the official Canvas assignment.

Use concise comments in completed code to document purpose, assumptions,
units, and non-obvious reasoning. Write requested explanations in Markdown
cells rather than Python comments.

Unless an assignment states otherwise, use `HW##_Firstname_Lastname.ipynb`
and `HW##_Firstname_Lastname.pdf`, replacing `##` with the two-digit
homework number.

## Homework 1: Python Basics and Basic Visualization

Apply foundational Python concepts to new, course-created manufacturing examples.
The assignment covers variables, types, unit conversion, list indexing and
slicing, dictionaries, conditions, loops, functions, Matplotlib, and concise
engineering interpretation. Estimated time: 60–90 minutes. Total: 100 points.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/WSU-AI-in-ME/ai-in-me-1/blob/main/homework/hw01_python_basics_and_visualization.ipynb)

## Homework 2: Scientific Computing, Manufacturing Signals, and Feature Engineering

Apply familiar methods to PHM milling data: physical time, Vibration Y features,
X/Y comparison, cut 1/315 FFT and STFT comparisons on shared scales, and exploratory
selection from 28 core features. Supplied processing and plotting code supports
short student calculations and engineering interpretation. Estimated time:
85 minutes (60–90 minute range). Total: 100 points.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/WSU-AI-in-ME/ai-in-me-1/blob/main/homework/hw02_scientific_computing_manufacturing_signals_and_feature_engineering.ipynb)

## Homework 3: Tool Wear Classification and Feature Choice

Choose three sensor features using training correlations and signal meaning,
then compare a Logistic Regression model with a supplied reference using Severe
F1, false positives and false negatives. Use Lab 5 as a code reference.
Estimated time: 45–60 minutes. Total: 100 points.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/WSU-AI-in-ME/ai-in-me-1/blob/main/homework/hw03_tool_wear_estimation_and_feature_comparison.ipynb)
