Wayne State University

This index links to student-facing homework starter files. Canvas and the
course syllabus are authoritative for assignment requirements, due dates,
rubrics, submission instructions, and course policies.

Coding homework will normally use a `.ipynb` starter notebook so that code,
figures, and short engineering responses remain in one reproducible file.
Supplemental `.md` files may explain setup, data provenance, or repository
structure, but they do not replace the official Canvas assignment.

Use concise comments in completed code to document purpose, assumptions,
units, and non-obvious reasoning. Write requested explanations in Markdown
cells rather than Python comments.

Unless an assignment states otherwise, use `HW##_Firstname_Lastname.ipynb`
and `HW##_Firstname_Lastname.pdf`, replacing `##` with the two-digit
homework number.

### Homework 1: Python Basics and Basic Visualization

slicing, dictionaries, conditions, loops, functions, Matplotlib, and concise
engineering interpretation. Estimated time: 60–90 minutes. Total: 100 points.
