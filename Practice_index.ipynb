## Week 1 — Course Introduction and Python Basics

### Practice 1: Python Fundamentals for Engineering

Work through runnable examples and varied **TRY IT** problem sets covering
variables, types, arithmetic, strings, conditions, lists, vector-like
sequences, list-based tables, dictionaries, loops, functions, and basic
Matplotlib plots. Problems range from foundation checks to short engineering
applications and challenges. No external data file is required.

## Week 2 — Scientific Computing Fundamentals

### Practice 2: Scientific Computing Fundamentals

Choose from 30 small exercises in six independent modules: arrays, shape/axis,
pandas and missing readings, physical time, mean/sample SD, and plots with
engineering interpretation. Use the 12-exercise short route or selected deeper
and debugging routes. Full-bank estimate: 90–150 minutes. Synthetic data are
defined in the notebook; no data download is needed.

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/WSU-AI-in-ME/ai-in-me-1/blob/main/practice/week02/practice02_scientific_computing.ipynb)

[Practice 2 Solution](https://colab.research.google.com/github/WSU-AI-in-ME/ai-in-me-1/blob/main/practice/week02/practice02_scientific_computing_solution.ipynb) — one possible approach, for self-checking after your attempt.

## Week 3 — Manufacturing Signals and Feature Engineering

### Practice 3: Manufacturing Signals and Feature Engineering

Choose from 25 small exercises in five independent modules: sampling and physical
time, time-domain features, FFT, FFT versus STFT, and feature relevance/redundancy.
Use the Short (10 exercises, 50–65 min), Standard (15, 80–100 min), Full
(25, 120–160 min), or Debugging (5, 35–50 min) route. Times are estimates;
complete selected exercises rather than every exercise in one sitting.
Synthetic examples require no data download. Optional PHM transfer T1 is separate
and disabled by default.
